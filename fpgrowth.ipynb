```python
import pandas  as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/Ataullha/CSE-436-Data-Mining-Lab/refs/heads/main/transaction_data.csv")
```

```python
df=df.fillna('*')
```

```python
df=df.drop('Transaction',axis=1)
```

```python
from mlxtend.preprocessing import TransactionEncoder
```

```python
te= TransactionEncoder()
```

```python
te_ary=te.fit_transform(df.values.tolist())
```

```python
te_ary=te_ary.tolist()
```

```python
df
```

```python
df.values.tolist()
```

```python
te_ary
```

```python
te.columns_
```

```python
transc_df=pd.DataFrame(te_ary,columns=te.columns_)
```

```python
transc_df
```

```python
transc_df=transc_df.drop('*',axis=1)
```

```python
transc_df
```

```python
#from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import fpgrowth
```

```python
 #rules=apriori(transc_df,min_support=.6,use_colnames=True)
```

```python
rules=fpgrowth(transc_df,min_support=.6,use_colnames=True)
```

```python
rules
```

```python
from mlxtend.frequent_patterns import association_rules
```

```python
assoc_rules=association_rules(rules,min_threshold=0.8)
```
